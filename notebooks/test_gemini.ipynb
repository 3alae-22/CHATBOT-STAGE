```python
import psycopg2
from pgvector.psycopg2 import register_vector
from dotenv import load_dotenv
from google import genai
from pathlib import Path
import os

env_path = Path(r"..\.env")
load_dotenv(env_path)

conn = psycopg2.connect(
    host="localhost", port=5432, dbname="db",
    user="db_user", password="db_password"
)
register_vector(conn)

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))

question = "Quelle est la durée de validité d'un certificat administratif ?"
question_embedding = client.models.embed_content(
    model="gemini-embedding-001",
    contents=question
).embeddings[0].values

with conn.cursor() as cursor:
    cursor.execute("""
        SELECT pdf_name, page_num, chunk_text, embedding <=> %s AS distance
        FROM dev.chunks
        ORDER BY distance
        LIMIT 5;
    """, (question_embedding,))
    results = cursor.fetchall()

for pdf_name, page_num, chunk_text, distance in results:
    print(f"[{pdf_name} - page {page_num}] (distance={distance:.4f})")
    print(chunk_text[:300])
    print("---")
```

```python
USER_PROMPT_TEMPLATE='''
Question:
{question}

Context:
{context}
'''
```

```python
def build_context(search_results):
    lines = []

    for i, row in enumerate(search_results, start=1):
        _, pdf_name, page_num, chunk_text, distance = row
        lines.append(f"[Source {i} — {pdf_name}, page {page_num}]")
        lines.append(chunk_text.strip())
        lines.append("")

    return "\n".join(lines).strip()
```

```python
def build_prompt(question, search_result):
    context=build_context(search_result)
    prompt=USER_PROMPT_TEMPLATE.format(
        question=question,
        context=context)
    return prompt.strip()
```
